```python
import pandas as pd
df = pd.read_csv("/content/session4_dummy_students.csv")
df['avg'] = df[['math', 'science', 'english']].mean(axis=1)
avg = df['avg'].to_numpy()
attendance = df['attendance_percent'].to_numpy()
city = df['city'].to_numpy()
top_students = df[avg > 85]
low_attendance = df[attendance < 75]
from_mathura = df[city == "Mathura"]

print(top_students)
print(low_attendance)
print(from_mathura)
```

```python
df['average'] = df[['math', 'science', 'english']].mean(axis=1)
df['result'] = np.where(df['average'] >= 40, 'Pass', 'Fail')

print("Class Average:", df['average'].mean())
print("Highest Average:", df['average'].max())
print("Lowest Average:", df['average'].min())

top_students = df[df['average'] == df['average'].max()]
print("Top Student:", top_students)

low_attendance = df[df['attendance_percent'] < 75]
print("Low Attendance:", low_attendance)

low_assignments = df[df['assignments_submitted'] < 6]
print("Low Assignments:", low_assignments)

df.to_csv("session4_student_report.csv", index=False)
```

```python
df["average"] = df[["math", "science", "english"]].mean(axis=1)
df["result"] = df["average"].apply(lambda x: "Pass" if x >= 40 else "Fail")

print("Class average:", df["average"].mean())
print("Highest average:", df["average"].max())
print("Lowest average:", df["average"].min())

top_avg = df["average"].max()
print("Top student(s):")
print(df[df["average"] == top_avg][["name", "average"]])

print("Students with attendance below 75:")
print(df[df["attendance_percent"] < 75][["name", "attendance_percent"]])

print("Students with fewer than 6 assignments:")
print(df.columns)
print(df[df["assignments_submitted"] < 6][["name", "assignments_submitted"]])

df.to_csv("session4_student_report.csv", index=False)
```
