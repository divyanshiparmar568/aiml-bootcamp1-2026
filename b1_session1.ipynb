```python
marks = [45, 60, 72, 38, 90]

total = 0

for i in range(len(marks) - 1):
    total += marks[i]

average = total / len(marks)

if average > 50:
    print("Class passed")
else:
    print("Class failed")
print("Average:",average)
```

```python
import pandas as pd
```

```python
data = {
    "student_id": [101, 102, 103, 104],
    "attendance": [75, 80, 23, 89],
    "quiz_score": [3, 5, 8, 10],
    "assignment_score": [20, 25, 15, 30]
}

df = pd.DataFrame(data)
```

```python
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df['total_score']=df['assignment_score']+df['quiz_score']
df   #add new feature
```

```python
df['eligible'] = (df['attendance'] > 75) & (df['quiz_score'] >= 5)
df  #filtering
```

```python
df['attendance'].mean() #aggregation=summarizing data
```

```python
df['quiz_score'] = df['quiz_score'] / 10  #transformation
df
```

```python
import matplotlib.pyplot as plt

plt.scatter(df['attendance'], df['quiz_score'])
plt.xlabel('Attendance')
plt.ylabel('Quiz Score')
plt.title('Attendance vs Quiz Score')
plt.show()
```

```python
df.plot(x='student_id', y='total_score', kind='bar', legend=False)
plt.title('Total Score by Student')
plt.ylabel('Total Score')
plt.show()
```
